```python
import numpy as np
```

```python
# Create two basic arrays
n1 = np.array([1, 2, 3])
n2 = np.array([4, 5, 6])

print(n1)
print(n2)
```

```python
# Define array and set its shape to 1x3
n1 = np.array([[22, 33, 44]])

print(n1)
print(n2)
```

```python
# Add [2, 2, 2] to n1
result = n1 + np.array([2, 2, 2])
print(result)
```

```python
# Subtract [2, 3, 4] from n1
result = n1 - np.array([2, 3, 4])
print(result)
```

```python
# Multiply n1 by 25
result = n1 * 25
print(result)
```

```python
# Perform element-wise multiplication
result = np.multiply(n1, [11, 12, 13])
print(result)
```

```python
# Reshape n1 to (3, 1)
# Create a 1x4 matrix filled with 3s
# Perform matrix multiplication using dot product
n1_reshaped = n1.reshape(3, 1)
threes = np.full((1, 4), 3)
result = np.dot(n1_reshaped, threes)
print(result)
```

```python
# Convert n1 to a 1D vector or handle transpose product
# Since n1 has shape (1, 3), np.dot(n1, n1.T) calculates the dot product
result = np.dot(n1, n1.T)
print(result)
```

```python
# Prime numbers between 1 and 23: [2, 3, 5, 7, 11, 13, 17, 19, 23]
# Compute inverse matrix
# Multiply v1 and v2, then round to handle floating-point precision
v1 = np.array([
    [2, 3, 5],
    [7, 11, 13],
    [17, 19, 23]
])
v2 = np.linalg.inv(v1)
identity = np.round(np.dot(v1, v2))
print(identity)
```

```python
# 1. Square root of all elements
print("Square Root:")
print(np.sqrt(v1))

# 2. Exponential (e^x) of all elements
print("\nExponential:")
print(np.exp(v1))

# 3. Sum of all elements
print("\nSum of elements:")
print(np.sum(v1))
```
